# 1 · Lab Bayesian optimization

The simplest abbot workflow: define a design space, pick an objective, and let
abbot propose experiments that climb toward the optimum. This is the "give a lab a
BO tool" case — a non-expert describes the knobs and the goal, nothing more.

The design space here is **heterogeneous**: a continuous temperature *and* a
categorical catalyst choice. (This mix is exactly what accelerator-oriented tools
coerce into float ranges — abbot keeps the catalyst discrete.)

```python
import torch
from abbot import (
    ProblemSpec, ContinuousDomain, EnumeratedDomain, ObjectiveSpec,
    SimpleBO, SingleTaskGPConstructor, run_campaign, SyntheticExperiment,
)
from abbot.acq_opt import DiscreteArgmax

torch.manual_seed(0)
```

### Define the problem
One continuous variable (`temp`) and one enumerated variable (`catalyst`), with a
single objective to **maximize** (`yield`).

```python
spec = ProblemSpec(
    domain=[
        ContinuousDomain(name="temp", lower=20.0, upper=80.0),
        EnumeratedDomain(name="catalyst", values=[0.0, 1.0, 2.0]),
    ],
    objectives=[ObjectiveSpec(name="yield", direction="MAXIMIZE")],
)
spec.variable_names
```

### A stand-in experiment
`SyntheticExperiment` is a seeded, smooth virtual lab — it stands in for a real
bench measurement so the notebook is self-contained. In practice you'd pass a
`ManualExperiment` (a human enters the measured value) or any object with a
`run(candidate) -> value` method.

```python
experiment = SyntheticExperiment(spec, seed=1)
```

### Run the campaign
`run_campaign` drives the propose → evaluate → retrain loop. Because the space is
all-enumerable in the catalyst dimension and continuous in temp, we use the
continuous optimizer for temp here by treating the enumerated values as a small
set — abbot picks the right candidate optimizer from the spec. We seed a few
space-filling points, then take 15 informed steps.

```python
import warnings
warnings.filterwarnings("ignore")  # quiet botorch's tiny-data GP-fit warnings

bounds = spec.bounds_tensor()
lo, hi = bounds[0], bounds[1]
seeds = lo + (hi - lo) * torch.rand(5, 2, dtype=torch.double)

result = run_campaign(
    experiment=experiment,
    spec=spec,
    policy=SimpleBO(optimizer=DiscreteArgmax()),  # enumerate the small mixed grid
    model_constructor=SingleTaskGPConstructor(bounds=bounds),
    evaluate=lambda obs: obs,          # the experiment already returns the objective
    initial_designs=seeds,
    n_shots=15,
)
```

### The recommendation
abbot hands back the denoised best design (posterior-mean argmax, not the luckiest noisy shot).

```python
best = result.best
print("recommended:", {n: round(float(v), 3) for n, v in zip(spec.variable_names, best.candidate.tolist())})
print("yield:", round(float(best.outputs), 3))
print("steps taken:", len(result.history))
```

### Convergence
Best objective seen so far, per step.

```python
import matplotlib.pyplot as plt
from abbot.analysis import convergence

ys = convergence(result.history, spec)
plt.figure(figsize=(6, 3.2))
plt.plot(range(len(ys)), ys, "-o", color="#4f46e5", ms=4)
plt.xlabel("step"); plt.ylabel("best yield so far"); plt.title("Lab BO convergence")
plt.tight_layout(); plt.show()
```

**What this showed:** a non-expert described a mixed continuous/categorical design
space and a single objective; abbot proposed experiments, learned a surrogate, and
recovered the best setting — no ML expertise required in the loop.
